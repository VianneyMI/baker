```python
import os
import certifi
import json
from pathlib import Path

from dotenv import load_dotenv
from pymongo import MongoClient
```

```python
load_dotenv()
```

```python
MONGODB_SERVER = os.environ.get("MONGODB_SERVER")
assert MONGODB_SERVER, "MONGODB_SERVER is not set"
```

```python
mongo_client = MongoClient(MONGODB_SERVER, tlsCAFile=certifi.where())
```

```python
db = mongo_client["baker"]
db.list_collection_names()
```

```python
path_to_data_folder = Path(os.getcwd()).parent / "data"
path_to_file_to_be_ingested = path_to_data_folder / "output" / "3_standardize_the_data" / "standardized_recipes.json"
```

```python
path_to_data_folder.exists(), path_to_file_to_be_ingested.exists()
```

```python
with open(path_to_file_to_be_ingested, "r") as file:
    recipes = json.load(file)
```

```python
recipes[:3]
```

```python
db.drop_collection("recipes")
db["recipes"].insert_many(recipes)
```
