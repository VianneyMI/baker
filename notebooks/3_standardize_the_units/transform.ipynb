### Mapping to standard units part I

```python
units_map = {
    "kilogram": "kg",
    "ml": "ml",
    "gram": "g",
    np.nan: "unit",
    "strip": "unit",
    "unit": "unit",
    "pound": "pound",
    "clove": "unit",
    "tablespoon": "unit",
    "ounce": "unit",
    "teaspoon": "unit",
    "dl": "dl",
    "cup": "unit",
    "can": "unit",
    "sprig": "unit",
    "slice": "unit",
    "dash": "unit",
    "milliliter": "ml",
    "bunch": "unit",
    "splash": "unit",
    "handful": "unit",
    "head": "unit",
    "block": "unit",
    "stick": "unit",
    "unknown": "unit",
    "g": "g",
    "leaf": "unit",
    "pinch": "unit",
    "bowl": "unit",
    "mug": "unit",
    "pint": "unit",
    "spoon": "unit",
    "package": "unit",
    "liter": "l",
    "inch": "unit",
    "glass": "unit",
    "jar": "unit",
    "centiliter": "cl",
    "twig": "unit",
    "sheet": "unit",
    "piece": "unit",
    "rib": "unit",
    "cube": "unit",
    "handfull": "unit",
    "bottle": "unit",
    "serving": "unit",
    "fluid_ounce": "unit",
    "recipe": "unit",
    "cm": "cm",
    "container": "unit",
    "bag": "unit",
    "shot": "unit",
    "loaf": "unit",
    "envelope": "unit",
    "l": "l",
    "medium": "unit",
    "large": "unit",
    "whole": "unit",
    "small": "unit",
    "percent": "unit",
    "pod": "unit",
    "fillet": "unit",
    "cl": "cl",
    "box": "unit",
    "quart": "unit",
    "strand": "unit",
    "stalk": "unit",
    "fruit": "unit",
}
```

```python
set(df["ingredients.original_unit"].unique().tolist()) == set(units_map.keys())
```

```python
df["ingredients.unit"] = df["ingredients.unit"].replace(units_map)
```

```python
df["ingredients.unit"].unique().tolist()
```

```python
# Conversion factors
conversion_factors = {
    "g": 0.001,  # grams to kilograms
    "ml": 0.001,  # milliliters to liters
    "cl": 0.01,  # centiliters to liters
    "dl": 0.1,  # deciliters to liters
    "pound": 0.453592,  # pounds to kilograms
    "cm": 1,  # centimeters to centimeters (no conversion needed)
    "l": 1,  # liters to liters (no conversion needed)
    "kg": 1,  # kilograms to kilograms (no conversion needed)
    "unit": 1,  # unit to unit (no conversion needed)
}

# Create a mapping for new units
new_units_map = {
    "g": "kg",
    "pound": "kg",
    "kg": "kg",
    "ml": "l",
    "cl": "l",
    "dl": "l",
    "l": "l",
    "cm": "cm",
    "unit": "unit",
}

# Replace quantities with converted quantities
df["ingredients.quantity"] = df["ingredients.quantity"] * df["ingredients.unit"].map(conversion_factors).fillna(1)
df["ingredients.quantity.min"] = df["ingredients.quantity.min"] * df["ingredients.unit"].map(conversion_factors).fillna(1)
df["ingredients.quantity.max"] = df["ingredients.quantity.max"] * df["ingredients.unit"].map(conversion_factors).fillna(1)

# Replace units with new units
df["ingredients.unit"] = df["ingredients.unit"].map(new_units_map).fillna(df["ingredients.unit"])

df.head()
```

### Fixing the quantities

```python
df[df["ingredients.quantity"].isna()].shape
```

```python
df[df["ingredients.quantity"].isna() & df["ingredients.quantity.min"].isna() & df["ingredients.quantity.max"].isna() ].shape
```

```python
df[df["ingredients.quantity"].isna() & df["ingredients.quantity.min"].isna() & df["ingredients.quantity.max"].isna() ][["id", "title", "ingredients.name", "ingredients.quantity", "ingredients.unit"]]
```

```python
selector = (df["ingredients.quantity"].isna() & df["ingredients.quantity.min"].isna() & df["ingredients.quantity.max"].isna())
df.loc[selector, "ingredients.quantity"] = df.loc[selector, "ingredients.quantity"].fillna(1)
df.head()
```

```python
df.loc[df.loc[:, "ingredients.quantity"].isna(), ["id","title", "ingredients.name", "ingredients.quantity", "ingredients.unit", "ingredients.quantity.min", "ingredients.quantity.max"]]
```

```python
df.loc[df.loc[:, "ingredients.quantity"].isna(), "ingredients.quantity"] = (df.loc[:, "ingredients.quantity.min"] + df.loc[:, "ingredients.quantity.max"]) / 2
df.loc[5,:]
```
