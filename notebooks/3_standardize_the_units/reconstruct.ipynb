```python
import ast
import os
from pathlib import Path

import pandas as pd
```

```python
path_to_data = Path(os.getcwd()).parents[1] / "data"
path_to_csv = path_to_data / "output" / "3_standardize_the_data" / "standardized_recipes.csv"
```

```python
df = pd.read_csv(path_to_csv)
df.head()
```

```python
# Define the columns to be included in the 'ingredient' dictionary
ingredient_columns = [
    'ingredients.id', 'ingredients.name', 'ingredients.quantity', 
    'ingredients.unit', 'ingredients.optional', "ingredients.original_unit",
    'ingredients.quantity.min', 'ingredients.quantity.max'
]

# Convert the selected columns to a dictionary with 'records' orientation
df['ingredient'] = df[ingredient_columns].rename(columns={
    'ingredients.id': 'id', 'ingredients.name': 'name', 
    'ingredients.quantity': 'quantity', 'ingredients.unit': 'unit',
    "ingredients.original_unit": "original_unit",
    'ingredients.optional': 'optional', 
    'ingredients.quantity.min': 'quantity_min', 
    'ingredients.quantity.max': 'quantity_max'
}).to_dict(orient='records')

# Display the first few rows of the updated DataFrame
df["ingredient"]
```

```python
df.head()
```

```python
df = df.drop(columns=ingredient_columns)
df.head()
```

```python
# Group by all columns except 'ingredient' and aggregate 'ingredient' back into a list
agg_map = {col: 'first' for col in df.columns if col != 'ingredient'} | {'ingredient': list}

df_reverted = df.groupby("id", as_index=False, dropna=False).agg(agg_map)
df_reverted.head()
```

```python
df_reverted.head()
```

```python
df_reverted.rename(columns={"ingredient": "ingredients"}, inplace=True)
```

```python
df_reverted["tags"] = df_reverted["tags"].apply(ast.literal_eval)
df_reverted["steps"] = df_reverted["steps"].apply(ast.literal_eval)
```

```python
path_to_output = path_to_data / "output" / "3_standardize_the_data" / "standardized_recipes.json"
df_reverted.to_json(path_to_output, orient='records', indent=4)
```
