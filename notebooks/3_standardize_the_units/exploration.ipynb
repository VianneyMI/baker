# Standardize the Units

```python
import os
from pathlib import Path

import pandas as pd
```

```python
path_to_data = Path(os.getcwd()).parents[1] / "data" 
path_to_recipes = path_to_data / "input" / "3_standardize_the_data" / "parsed_recipes.json"
```

```python
df = pd.read_json(path_to_recipes)
df.head()
```

```python
# Creating a ID for the recipe

df.reset_index(inplace=True)
df.rename(columns={"index":"id"}, inplace=True)
df.head()
```

```python
_df = df.copy(deep=True)
```

```python
_df.shape
```

## Exploring the ingredients

```python
df = _df.explode("ingredients")
df.head()
```

```python
dfi = pd.json_normalize(df["ingredients"])
dfi.head()
```

```python
dfi.columns = map(lambda v:f"ingredients.{v}", dfi.columns)
dfi.head()
```

```python
df:pd.DataFrame = df.drop("ingredients", axis=1).reset_index(drop=True).join(dfi)
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
# Reorder columns
new_column_order = [
    "id",
    "title",
    "serving_size",
    "preparation_time",
    "cooking_time",
    "ingredients.id", 
    "ingredients.name",
    "ingredients.quantity", 
    "ingredients.unit", 
    "ingredients.optional",
    "ingredients.quantity.min", 
    "ingredients.quantity.max",
    "date",
    "tags",
    "steps",
    "ingredients_source_text",
    "directions_source_text"

]

df = df[new_column_order]
```

```python
df["ingredients.unit"].unique().tolist()
```

Some units have unexpecte values like None or nan. We will explore the ingredients to see what we can do to standardize the units.
In particular, we need to check if this is a LLM parsing problem or if the issue occured during post-processing.

```python
pd.set_option("display.max_colwidth",None)
```

```python
ingredients_without_unit = df[df["ingredients.unit"].isna()][["id", "title", "ingredients.name", "ingredients.quantity", "ingredients.unit", "ingredients.optional", "ingredients_source_text"]]
ingredients_without_unit
```

It looks like the ingredients without unit are mainly ingredients whose unit were not present in the original recipe. 
Moreover, it looks like it's mainly optional spicies

```python
ingredients_without_unit["id"].value_counts()
```

"Only" 69 recipes are concerned by this issue.
Might be worth considering dropping those recipes

```python
ingredients_without_unit.shape
```

```python
ingredients_without_unit[~ingredients_without_unit["ingredients.optional"]]
```

My assumption around the ingredients without unit being mainly optional spices is wrong.
However, it is correct to say the unit is missing when it was already missing in the original dataset

```python
ingredients_without_unit[ingredients_without_unit["id"].isin([236,249])]
```

```python
ingredients_with_na_unit = df[df["ingredients.unit"] == "N/A"][["id", "title", "ingredients.name", "ingredients.quantity", "ingredients.unit", "ingredients.optional", "ingredients_source_text"]]
ingredients_with_na_unit
```
